```python
import string
import nltk
import spacy

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from collections import Counter
```

```python
text = """
Apple is looking at buying a startup in the United Kingdom for $1 billion.
The company was founded by John in 2019.
He loves running, playing football and reading books.
"""
```

```python
tokens = word_tokenize(text)

print(tokens)
```

```python
tokens = [word.lower() for word in tokens]

print(tokens)
```

```python
tokens = [word for word in tokens if word not in string.punctuation]

print(tokens)
```

```python
stop_words = set(stopwords.words("english"))

filtered = [word for word in tokens if word not in stop_words]

print(filtered)
```

```python
stemmer = PorterStemmer()

stemmed = [stemmer.stem(word) for word in filtered]

print(stemmed)
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized = [lemmatizer.lemmatize(word) for word in filtered]

print(lemmatized)
```

```python
frequency = Counter(lemmatized)

print(frequency)

print("\nTop 10 Words")

for word, count in frequency.most_common(10):
    print(word, ":", count)
```

```python
pos = nltk.pos_tag(lemmatized)

print(pos)
```

```python
nlp = spacy.load("en_core_web_sm")

doc = nlp(text)

for entity in doc.ents:
    print(entity.text, "---->", entity.label_)
```
