```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = [
    "running", "walked", "studies", "better", "best",
    "went", "gone", "eating", "ate", "written",
    "writing", "children", "mice", "feet", "cars",
    "leaves", "playing", "bought", "driving", "seen"
]

for word in words:
    print(f"{word} -> {lemmatizer.lemmatize(word, pos='v')}")
```
